```python

df = pd.read_csv("bedtime_screentime_sleep_debt.csv")

print("\n--- Кількість унікальних значень у категоріальних стовпцях ---")
# Знаходимо лише текстові/категоріальні колонки (тип 'object' або 'str')
categorical_cols = df.select_dtypes(include=["object", "str"]).columns

# Виводимо кількість унікальних значень для кожного такого стовпчика
for col in categorical_cols:
  print(f"{col}: {df[col].nunique()} унікальних значень")
```

```python
#Для двох числових стовпців обчисліть і запишіть: кількість, мінімум, максимум, розмах, середнє, медіану, моду, стандартне відхилення (std). 
#Для аналізу я вибeру дві колонки, які найкраще відображає вплив гаджетів на сон:
#bedtime_phone_minutes — час (у хвилинах), який користувач проводить із телефоном перед сном.
#sleep_latency_min — час засинання .

df = pd.read_csv("bedtime_screentime_sleep_debt.csv")
    # Вибираємо потрібні стовпці та застосовуємо функції агрегації
cols = ["bedtime_phone_minutes", "sleep_latency_min"]
result_table = df[cols].agg(['count', 'min', 'max', 'mean', 'median', 'std'])
result_table.loc['mode'] = [df[col].mode().iloc[0] for col in cols]

print(result_table) 

 
```

```python
#Порахуйте кореляцію для однієї пари числових стовпців (не беріть стовпці-ідентифікатори). Одним-двома реченнями поясніть, що є «типовим»,

correlation = df['bedtime_phone_minutes'].corr(df['sleep_latency_min'])
print(f"Кореляція між телефоном перед сном і часом засинання: {correlation:.2f}")
```

```python
# bedtime_phone_minutes (Час із телефоном):Середнє значення — 59.25 хв, але стандартне відхилення (std) становить 36.93 хв.
# Це величезний розкид: дані сильно «розмазані» навколо середнього. Це означає, що у вибірці є як «мінімалісти» 
# (хто вимикає телефон за 1 хвилину до сну), так і завзяті користувачі, які проводять у гаджетах до 3 годин (180 хв)

# sleep_latency_min (Час засинання): Середнє — 40.67 хв, а стандартне відхилення — 17.44 хв.
# Розкид тут помірніший, але мінімум становить 6 хвилин, а максимум аж 123.3 хв (понад 2 години). 
# Такий довгий хвіст максимальних значень чітко показує наявність групи людей із вираженим безсонням або проблемами із засинанням.

# Кореляція між часом користування телефоном перед сном і часом засинааня 0,86 
#  швидше за все, це означає статистичний зв'язок: люди, які довше дивляться в телефон, у середньому довше засинають.
# але ми не можемо достовірно ще казати про пряму залежність користування телефоном перед сном, на це можуть впливати інші чинники:
#  перевтома, стрес...
```

```python
# Ризик: Підозрілі викиди / аномалії в числових даних (sleep_latency_min)
# Опис: Екстремальні значення (викиди), які виходять за межі здорового глузду або фізіологічної норми 
# (наприклад, час засинання понад 2 години). Вони можуть спотворювати середні показники та псувати якість моделей.

# Виділяємо аномальні значення засинання (більше 120 хвилин)
outliers = df[df["sleep_latency_min"] > 120]
print(
    f"Кількість записів з аномально довгим засинанням (>120 хв):"
    f" {len(outliers)}"
)
print("Максимальне зафіксоване значення:", df["sleep_latency_min"].max())
```
